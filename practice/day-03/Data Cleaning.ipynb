```python
"""
File: Data Cleaning.ipynb
Author: Abasin Khan
Date: 2026-09-06

Description:
This script demonstrates how to clean and prepare data using Pandas.
It handles missing values, removes inconsistencies, and improves data quality.
"""
import pandas as pd
df = pd.read_excel("pandas_data_cleaning_practice.xlsx")
 
# Always explore data before cleaning it.
df.head()
df.info()
df.shape
df.columns
```

```python
# Checking Missing Values 
df.isnull().sum() # Count Missing Values
```

```python
# Removing Missing Values
df.dropna() # Remove rows containing missing values
```

```python
# Filling Missing Values
df["Age "] = df["Age "].fillna(0) # Filling with a Specific Value
df[" City"] = df[" City"].fillna("Unknown")

df = df.fillna(0) # Fill All Missing Values.
df["Age "] = pd.to_numeric(df["Age "], errors="coerce")
df
```

```python
# Detecting Duplicate Rows and Removing it
df.duplicated()
df.duplicated().sum() # Counting it
df = df.drop_duplicates() # Removing Duplicate Rows
df
```

```python
# Removing Extra spaces

df.columns = df.columns.str.strip()
df.columns = df.columns.str.lower() # Convert to Lower
df.columns = df.columns.str.replace(" ", "_") # Replacing Spaces with Underscores
df
```

```python
# Changing DataTypes
df.dtypes # Checking Data Types
df["age"] = df["age"].astype(int) # Convert to Integer
df["email"] = df["email"].astype(str)
df
```

```python
# Working with Dates
df["joining_date"] = pd.to_datetime(
    df["joining_date"],
    errors="coerce"
)
df["joining_date"].isna().sum()  # Check for Missing dates
```

```python
# Finding Unique Values
df["city"] = df["city"].str.strip().str.title()
df["city"].unique()
df["city"].nunique() # Count Unique Values
```

```python
# Replacing Incorrect Values
df["gender"] = df["gender"].replace({
    "M": "Male",
    "male": "Male",
    "MALE": "Male"
})
```

```python
# Removing Invalid Data
df = df[
    (df["age"] >= 0) &
    (df["age"] <= 100)
]
```

```python
# Removing Unnecessary Column
df = df.drop(columns=["unnecessary_column"])
```

```python
# Reseting Index
df = df.reset_index(drop=True)
```
