```python
import pandas as pd
df = pd.read_excel("Students_Data.xlsx",sheet_name=0)

# Look at the Data:
df.head() # By Default head() gives first 5 rows. 
df.tail() # tail() gives last 5 rows by default.
df.sample(5)
```

```python
# Understand the structure
df.shape # shape--> Check the size of Data (rows,column)
df.shape[0] # Get only rows
df.shape[1] # Get only columns
df.columns
df.index
df.size
```

```python
# Understand data types
df.info() # Get Complete Information
df.dtypes
len(df) # Returns the total number of Rows
```

```python
# Check missing data
df.isnull.sum().sum()
df.isnull().sum() # isnull() is used to Check missing values sum() is used to give total missing values
df.count() # --> Count Non-Missing Values
```

```python
df.notnull().sum() # Check for Non-Missing Values then counting them
df.describe() #--> Statistical Summary: i.e count, mean, std, min.. etc
df.dtypes #--> Check Data Types
```

```python
# Working with Columns

df.columns #columns--> Get column Names
df.columns.tolist() # tolist--> To convert column names into a normal Python list
df["Name"] # View one column 
df.Name # We can also use this (Same Result)
df[["Name", "Age"]] # --> View Multiple Columns
df["Name"].unique() # Check Unique values in particular column
```

```python
# Working with rows

df.duplicated() #--> Check for Duplicate Rows (dtype: Bool)
df[df.duplicated()] # See Duplicate Rows
df.iloc[2] # Select Specific Row
df.loc[0:5, ["Name", "Age"]] # Select Rows and Columns 
df.sample(5) # Show Random Rows
```

```python
# Statistics 

df["Age"].min() #--> Find Minimum Value.
df["Age"].max() #--> Find Maximum Value.
df["Age"].mean() # Calculate Average.
df["Age"].median() # Calculate Median.
df["Age"].mode() # Find most Common Value.
df["Age"].std() # Shows how spread out the values are.

```

```python
# Checking Data Quality
df.duplicated().sum()
df[df.duplicated()]
```
