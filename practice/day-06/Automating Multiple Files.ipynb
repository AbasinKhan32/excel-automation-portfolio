```python
# os is a built-in Python module used to interact with your computer's operating system
import os
print(os.getcwd()) # getcwd() means: Get Current Working Directory
files = os.listdir("excel files") # This shows the files and folders inside a directory

print(files)
```

```python
# glob allows us to find files based on a pattern.
import glob
files = glob.glob("excel files/*.xlsx") # Find every .xlsx file inside the excel files folder

print(files)
```

```python
# Combine glob with Pandas
import pandas as pd
files = glob.glob("excel files/*.xlsx")

for file in files:
    df = pd.read_excel(file)

    print(file)
    print(df)
```
