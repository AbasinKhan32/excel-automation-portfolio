```python
import pandas as pd

# Read Excel file
df = pd.read_excel("pandas_filtering_sorting_practice.xlsx")

# Filter employees
filtered_df = df[
    (df["Department"] == "IT") &
    (df["Salary"] >= 50000)
]

# Sort by salary
result = filtered_df.sort_values(
    "Salary",
    ascending=False
)

# Reset index
result = result.reset_index(drop=True)

# Display result
print(result)

# Save result to new Excel file
result.to_excel(
    "it_employees_sorted.xlsx",
    index=False
)
```
