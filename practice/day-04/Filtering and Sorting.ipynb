```python
"""
File: Filtering and Sorting.ipynb
Author: Abasin Khan
Date: 2026-09-07

Description:
This script demonstrates how to filter and sort data using Pandas.
It helps select specific records and arrange data based on column values.
"""

import pandas as pd
df = pd.read_excel("pandas_filtering_sorting_practice.xlsx")

```

```python
'''
What is Filtering?
Filtering means selecting only the rows that match a specific condition.
For example, imagine you have 1,000 employees but only want employees from the IT department
'''
it_employees = df[df["Department"] == "IT"]

print(it_employees)
```

```python
# Filter Using One Condition
result = df[df["Age"] > 30] # Employees older than 30
result=df[df["Salary"] > 60000]
result=df[df["Experience_Years"] >= 5]
result=df[df["Department"] != "IT"]
print(result)
```

```python
# Filter Text/String Values
df[df["City"] == "Abbottabad"]
df[df["Performance"] == "Excellent"]
df[df["Department"] == "IT"]
```

```python
# Filtering with Multiple Conditions: AND, OR, NOT
# AND
result = df[
    (df["Department"] == "IT") & 
    (df["Salary"] > 60000)
]

print(result) 
```

```python
# OR
result = df[
    (df["Department"] == "IT") |
    (df["Department"] == "Finance")
]

print(result)
```

```python
# Filter Using isin() --> When you want to filter multiple values from the same column
result = df[
    df["Department"].isin(["IT", "Finance"])
]

print(result)
```

```python
# Filter Using between()
result = df[
    df["Age"].between(25, 30) # Employees age between 25 and 30 
]
print(result)


result = df[
    df["Salary"].between(50000, 70000)  # Salary Between 50000 and 70000
]
print(result)
```

```python
# Filter Missing Values
result = df[df["Salary"].isna()] # Find Missing Values
result = df[df["Salary"].notna()] # Find Rows that are not Missing
```

```python
# Filter Text Using str.contains()
# Names containing Khan
result = df[
    df["Name"].str.contains("Khan", case=False, na=False) 
]

print(result)
```

```python
# Filter and Select Specific Columns
# Show the Name and Salary of IT employees.
result = df.loc[
    df["Department"] == "IT", 
    ["Name", "Salary"]
]

print(result)
```

```python
'''
What is Sorting?
Sorting means arranging your data in a specific order.
'''
sorted_df = df.sort_values("Salary")

```

```python
# Sort One Column 
sorted_df = df.sort_values("Salary")

print(sorted_df)

# Sort High to Low 
sorted_df = df.sort_values(
    "Salary",
    ascending=False
)

print(sorted_df)
```

```python
# Sort Names alphabetically
sorted_df = df.sort_values("Name")

print(sorted_df)

# Reverse
sorted_df = df.sort_values(
    "Name",
    ascending=False
)
```

```python
# Sort Multiple Columns 
sorted_df = df.sort_values(
    by=["Department", "Salary"],
    ascending=[True, False]
)

```

```python
# Filter and Sort Together 
result = df[
    df["Department"] == "IT"
].sort_values(
    "Salary",
    ascending=False
)

print(result)
```

```python
# Using query() for Filtering
result = df.query(
    'Department == "IT"'
)
print(result)

# Multiple Conditions
result = df.query(
    'Age > 25 and Salary > 60000'
)
```

```python
# Reset Index after Sorting or Filtering
result = df[
    df["Department"] == "IT"
].sort_values(
    "Salary",
    ascending=False
).reset_index(drop=True)
```

```python
# Save Filtered/Sorted Data back to Excel
result.to_excel(
    "filtered_employees.xlsx",
    index=False
)
```
