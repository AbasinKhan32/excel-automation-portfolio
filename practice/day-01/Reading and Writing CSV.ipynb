```python
"""
File: Reading and Writing CSV.py
Author: Abasin Khan
Date: 2026-09-04

Description:
This script demonstrates how to read data from CSV files using Pandas. It also shows how to write and save data to CSV files.
"""
import pandas as pd

# 'Header=None' is used for when your CSV file does not contain column names (headers).
df = pd.read_csv("sample_data.csv",header=None, names=["stocks","Mobile","Cars","Laptops","Bags"])

# Here 'nrows=3' tells Pandas how many rows of the CSV file to read.
df = pd.read_csv("sample_data.csv",nrows=3)

# na_values tells Pandas which values should be considered missing values (NaN).
df = pd.read_csv("sample_data.csv",na_values={
    "Age":["not available","n.a.",-8],
    "Department":["not available","n.a."],
    "Score":["not available","n.a."],
})

print(df)
```

```python
# index=False means not to give index in new file.
df.to_csv("New.csv",index=False)
```

```python
# .columns is used to view column names of a DataFrame
df.columns
```

```python
# This line will create new csv file with only "Name" & "Age" Columns
df.to_csv("2_columns.csv",columns=["Name","Age"],index=False)
```
